## LRFM Analysis Implementation

```python

# Calculate the reference date (max order date + 1 day)
reference_date = df['Order Date'].max() + pd.Timedelta(days=1)

# Group by customer to calculate LRFM metrics
customer_data = df.groupby('Customer Name').agg({
    'Order Date': [
        ('recency', lambda x: (reference_date - x.max()).days),  # Recency: days since last purchase
        ('length', lambda x: (x.max() - x.min()).days)          # Length: days between first and last purchase
    ],
    'Order ID': [
        ('frequency', 'nunique')                                # Frequency: number of unique orders
    ],
    'Sales': [
        ('monetary', 'sum')                                     # Monetary: total sales amount
    ]
}).reset_index()

# Flatten the multi-index columns
customer_data.columns = ['Customer Name', 'recency', 'length', 'frequency', 'monetary']

# Calculate LRFM scores (quantile-based scoring)
def calculate_rfm_scores(df, column, ascending=True):
    # Higher score is better (1=lowest, 4=highest)
    return pd.qcut(df[column], q=4, labels=[1, 2, 3, 4], duplicates='drop')

customer_data['recency_score'] = calculate_rfm_scores(customer_data, 'recency', ascending=False)
customer_data['length_score'] = calculate_rfm_scores(customer_data, 'length')
customer_data['frequency_score'] = calculate_rfm_scores(customer_data, 'frequency')
customer_data['monetary_score'] = calculate_rfm_scores(customer_data, 'monetary')

# Calculate overall LRFM score (average of all dimensions)
customer_data['lrfm_score'] = (customer_data['recency_score'].astype(int) + 
                              customer_data['length_score'].astype(int) + 
                              customer_data['frequency_score'].astype(int) + 
                              customer_data['monetary_score'].astype(int)) / 4

# Create LRFM segments
def create_lrfm_segments(row):
    if row['lrfm_score'] >= 3.5:
        return 'Loyal Customers'
    elif row['lrfm_score'] >= 3.0:
        return 'Potential Loyalists'
    elif row['lrfm_score'] >= 2.5:
        return 'Inertia Loyalists'
    elif row['lrfm_score'] >= 2.0:
        return 'Regular Customers'
    elif row['lrfm_score'] >= 1.5:
        return 'Recent Customers'
    elif row['lrfm_score'] >= 1.0:
        return 'New Customers'
    else:
        return 'Passive Customers'

customer_data['lrfm_segment'] = customer_data.apply(create_lrfm_segments, axis=1)

# Display the results
print("Customer LRFM Analysis Results:")
display(customer_data.head())

# Visualize the segments
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.countplot(data=customer_data, y='lrfm_segment', order=customer_data['lrfm_segment'].value_counts().index)
plt.title('Customer Segments Distribution')
plt.xlabel('Number of Customers')
plt.ylabel('LRFM Segment')
plt.show()

# Analyze segment characteristics
segment_analysis = customer_data.groupby('lrfm_segment').agg({
    'recency': 'mean',
    'length': 'mean',
    'frequency': 'mean',
    'monetary': 'mean',
    'Customer Name': 'count'
}).rename(columns={'Customer Name': 'count'}).sort_values('count', ascending=False)

print("\nSegment Characteristics:")
display(segment_analysis)
```

# Customer Segment

## Loyal Customer
someone who consistently chooses a specific brand or product over competitors, often due to a positive experience and a strong emotional connection.

## Potential Loyalist
a customer who has made a few purchases and shows signs of developing loyalty, but hasn't yet become a consistently loyal customer.

## Intertia Loyalist
describes customers who continue to purchase a product or service not because of genuine loyalty or satisfaction, but due to habit and ease of sticking with the familiar.

## Regular Customer
someone who habitually buys products or services from a particular business or person frequently.

## Recent Customer
are individuals who have made their most recent purchase in the shortest time frame relative to the reference date of the analysis.

## New Customer
are those who have just started their relationship with your business. They have made their first purchase recently, and their engagement level is still in its early stage.

## Passive Customer
are those who are still part of your customer base but show signs of declining activity. They haven't made a recent purchase, and their engagement is neither high nor completely inactive.

*K-Cluster analysis*

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
# Clustering K-Means
features = customer_data[['recency', 'length', 'frequency', 'monetary']]
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

# Terapkan K-Means clustering
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
customer_data['Cluster'] = kmeans.fit_predict(features_scaled)
customer_data
```

```python
cluster_stats = customer_data.groupby('Cluster')[['recency', 'length', 'frequency', 'monetary']].mean().round(1)
cluster_counts = customer_data['Cluster'].value_counts().rename_axis('Cluster').reset_index(name='Customer Count')
cluster_summary = pd.merge(cluster_counts, cluster_stats, on='Cluster')
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=cluster_summary, x='Cluster', y='Customer Count', palette='pastel')
plt.title('Total Customer per Cluster (K-Means)', fontsize=14, fontweight='bold')
plt.xlabel('Cluster')
plt.ylabel('Total Customer')
plt.tight_layout()
plt.show()
```

```python
# Visualize the average LRFM metrics for each cluster
plt.figure(figsize=(10, 6))
sns.heatmap(cluster_summary.set_index('Cluster')[['recency', 'length', 'frequency', 'monetary']],
            annot=True, fmt=".1f", cmap="YlOrRd", linewidths=.5)
plt.title('Heatmap Rata-Rata LRFM per Cluster (K-Means)', fontsize=14, fontweight='bold')
plt.xlabel('Metrik LRFM')
plt.ylabel('Cluster')
plt.tight_layout()
plt.show()
```

-Initiate mass reactivation campaigns aimed at reminding them of your offerings.

-If reactivation is unsuccessful, prioritize other segments that are more active.

-Use the insights from this group to refine retention strategies and prevent churn.
